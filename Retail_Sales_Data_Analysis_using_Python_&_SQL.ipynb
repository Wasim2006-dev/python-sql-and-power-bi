#import sql

```python
import pandas as pd
import sqlite3
```

```python
df=pd.read_csv('sales_store_updated_allign_with_video.csv')
df
```

```python
conn = sqlite3.connect("sales_store.db")

df.to_sql(
    "sales",
    conn,
    if_exists="replace",
    index=False
)

print("Data successfully inserted into SQLite!")
```

```python
query = "SELECT * FROM sales LIMIT 10"

result = pd.read_sql_query(query, conn)

result
```

```python
pd.read_sql_query(
    "SELECT COUNT(*) AS total_records FROM sales",
    conn
)
```

#Category-wise sales quantity:

```python
query = """
SELECT product_category,
       SUM(quantiy) AS total_quantity
FROM sales
GROUP BY product_category
ORDER BY total_quantity DESC;
"""

pd.read_sql_query(query, conn)
```

#Payment mode analysis:

```python
query = """
SELECT payment_mode,
       COUNT(*) AS transactions
FROM sales
GROUP BY payment_mode
ORDER BY transactions DESC;
"""

pd.read_sql_query(query, conn)
```

#Status-wise transactions:

```python
query = """
SELECT status,
       COUNT(*) AS total_transactions
FROM sales
GROUP BY status;
"""

pd.read_sql_query(query, conn)
```

#Top 10 products:

```python
query = """
SELECT product_name,
       SUM(quantiy) AS total_quantity
FROM sales
GROUP BY product_name
ORDER BY total_quantity DESC
LIMIT 10;
"""

pd.read_sql_query(query, conn)
```
