#connect sql

```python
import pandas as pd
import sqlite3
```

```python
df=pd.read_csv('cars data.csv')
df
```

#create sqlite database

```python
conn = sqlite3.connect("cars.db")
```

#upload database in sqlite

```python
df.to_sql(
    "cars",
    conn,
    if_exists="replace",
    index=False
)
```

#check data is uploaded in sqlite

```python
query = """
SELECT *
FROM cars
LIMIT 10;
"""

pd.read_sql_query(query, conn)
```

#total recods

```python
pd.read_sql_query(
    "SELECT COUNT(*) AS total_records FROM cars;",
    conn
)
```

#check sqlite table

```python
pd.read_sql_query(
    "PRAGMA table_info(cars);",
    conn
)
```

#unique cars manufacturer

```python
pd.read_sql_query("""
SELECT COUNT(DISTINCT Make) AS Total_Makes
FROM cars;
""", conn)
```

#cars by manufacturer

```python
pd.read_sql_query("""
SELECT
    Make,
    COUNT(*) AS Total_Cars
FROM cars
GROUP BY Make
ORDER BY Total_Cars DESC;
""", conn)
```

#Cars by Type

```python
pd.read_sql_query("""
SELECT
    Type,
    COUNT(*) AS Total_Cars
FROM cars
GROUP BY Type
ORDER BY Total_Cars DESC;
""", conn)
```

#Cars by Origin

```python
pd.read_sql_query("""
SELECT
    Origin,
    COUNT(*) AS Total_Cars
FROM cars
GROUP BY Origin
ORDER BY Total_Cars DESC;
""", conn)
```

#drive train analysis

```python
pd.read_sql_query("""
SELECT
    DriveTrain,
    COUNT(*) AS Total_Cars
FROM cars
GROUP BY DriveTrain
ORDER BY Total_Cars DESC;
""", conn)
```

#Average Horsepower

```python
pd.read_sql_query("""
SELECT
    AVG(Horsepower) AS Avg_Horsepower
FROM cars;
""", conn)
```

#Maximum Horsepower cars

```python
pd.read_sql_query("""
SELECT
    Make,
    Model,
    Horsepower
FROM cars
ORDER BY Horsepower DESC
LIMIT 10;
""", conn)
```

#Minimum Horsepower cars

```python
pd.read_sql_query("""
SELECT
    Make,
    Model,
    Horsepower
FROM cars
ORDER BY Horsepower ASC
LIMIT 10;
""", conn)
```

#Average MPG

```python
pd.read_sql_query("""
SELECT
    AVG(MPG_City) AS Avg_City_MPG,
    AVG(MPG_Highway) AS Avg_Highway_MPG
FROM cars;
""", conn)
```

#Best fuel-efficient cars

```python
pd.read_sql_query("""
SELECT
    Make,
    Model,
    MPG_City,
    MPG_Highway
FROM cars
ORDER BY MPG_Highway DESC
LIMIT 10;
""", conn)
```

#Average price — MSRP

```python
pd.read_sql_query("""
SELECT
    AVG(CAST(REPLACE(TRIM(MSRP), '$', '') AS REAL)) AS Average_MSRP
FROM cars;
""", conn)
```

#Most expensive cars

```python
pd.read_sql_query("""
SELECT
    Make,
    Model,
    MSRP
FROM cars
ORDER BY CAST(REPLACE(TRIM(MSRP), '$', '') AS REAL) DESC
LIMIT 10;
""", conn)
```

#Cheapest cars

```python
pd.read_sql_query("""
SELECT
    Make,
    Model,
    MSRP
FROM cars
ORDER BY CAST(REPLACE(TRIM(MSRP), '$', '') AS REAL) ASC
LIMIT 10;
""", conn)
```

#Average price by manufacturer

```python
pd.read_sql_query("""
SELECT
    Make,
    COUNT(*) AS Total_Cars,
    ROUND(AVG(CAST(REPLACE(TRIM(MSRP), '$', '') AS REAL)), 2) AS Avg_MSRP
FROM cars
GROUP BY Make
ORDER BY Avg_MSRP DESC;
""", conn)
```

#Average horsepower by Type

```python
pd.read_sql_query("""
SELECT
    Type,
    COUNT(*) AS Total_Cars,
    ROUND(AVG(Horsepower), 2) AS Avg_Horsepower
FROM cars
GROUP BY Type
ORDER BY Avg_Horsepower DESC;
""", conn)
```

#Top 10 heaviest cars

```python
pd.read_sql_query("""
SELECT
    Make,
    Model,
    Weight
FROM cars
ORDER BY Weight DESC
LIMIT 10;
""", conn)
```

#Database close

```python
conn.close()
```
